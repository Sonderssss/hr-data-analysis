```python
import pandas as pd

df = pd.read_parquet('../data/processed/hr_data.parquet')

df.head()
```

##### Company Attrition Rate

39.2%, 392/1000 employees left the company 

```python
total_employees = df.shape[0]

employees_left = df[df['Attrition']=='Yes'].shape[0]

attrition_rate = (employees_left / total_employees) * 100

attrition_rate
```

##### Department

```python
dpt_counts = pd.crosstab( df['Department'], df['Attrition'])

# Pick the ones that left(Yes)
dpt_counts

dpt_attrition = (pd.crosstab( df['Department'], df['Attrition'], normalize='index')['Yes'] * 100 ).round(2).sort_values(ascending=False)

# Used value counts cause taking the 'All' column adds totals and theres a missing value
print(pd.concat(
        [df['Department'].value_counts(), dpt_counts['Yes'], dpt_attrition], 
        axis=1, 
        keys=['Counts','Left','Attrition']
        ).sort_values(by='Attrition' , ascending=False).reset_index()
     )
```

Strong evidence

Job Satisfaction: a 33-point gap with a p-value of 0.0. Scores 1 and 2 leave far more than 3 to 5. It's the clear winner.

Joining Year: worth a closer look
Employees who joined in 2018 left more often (48%) than those who joined in 2020 (36%).
The gap is probably real, but there are two reasons to be careful:
People who joined in 2018 have simply had more years in which they could leave.
Because we checked ten factors, one of them can look real by luck alone.
This is a lead worth following up, not a firm conclusions
