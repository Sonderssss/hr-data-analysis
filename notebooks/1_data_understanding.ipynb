#### Messy values

Age: it’s stored as text (one value is “thirty”) and 159 are missing. Birth Year is complete, so you can work age out from it.
Salary: it’s stored as text, with entries like “SIXTY THOUSAND”.
Department and Employment Status: the capitalisation is inconsistent (finance / Finance, contract / Contract, hr / HR).
Benefits Enrolled: it mixes YES / Yes / N / no, plus blanks.
Dates: Joining Date and Last Promotion Date are both text. They also come in mixed formats, including 2024.08.12.
