```python
import pandas as pd

df = pd.read_csv('../data/raw/hr_data.csv')

df.head(20)
```

### How to fix the columns 

##### Department 

```python
# Department
department_map = {
   'Finance':'Finance',
    'Sales':'Sales',
    'HR':'HR',
    'IT':'IT',
    'Marketing':'Marketing',
    'marketing':'Marketing',
    'finance':'Finance',  
    'it':'IT',
    'hr':'HR',     
    'sales':'Sales'
}

df['Department'] = df['Department'].str.strip().map(department_map)

df['Department'].value_counts()
```

##### Employment Status

```python
## Employment Status 
employment_status_map = {
    'Full-Time':'Full-Time',
    'Part-Time':'Part-Time',
    'full-time':'Full-Time',
    'part-time':'Part-Time',
    'FULL-TIME':'Full-Time',
    'PART-TIME':'Part-Time',
    'contract':'Contract',
    'Contract':'Contract',
}

df['Employment Status'] = df['Employment Status'].str.strip().map(employment_status_map)


df['Employment Status'].value_counts()
```

##### Age

```python
# Fixing the Age column
print(df['Age'].dtype)

#replacing all the string values in the Age column with the numeric values
df['Age'] = df['Age'].replace('thirty', 30)

# Int64 is a nullable integer data type in pandas that allows for missing values (NaN) in the column.
# It will not raise errors when encountering NaN values.
df['Age'] = pd.to_numeric(df['Age'], errors='coerce').astype('Int64')

# Confirm the data type of the age column after the conversion
print(df['Age'].dtype)

# Calculating the age based on the birth year and 2024 as the current year.
df['Age'] = df['Birth Year'].apply(lambda x: 2024 - x if pd.notnull(x) else x)

df['Age'].head(10)

# Confirmaing statistics of the Age column after the conversion
df['Age'].agg(['min', 'max', 'mean', 'median'])
```

##### Salary


```python
# Fixing the salary column

# Removing any non-numeric characters from the salary column and converting it to a numeric data type
df['Salary'] = df['Salary'].replace('SIXTY THOUSAND', 60000)

df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce').astype('Int64')

df['Salary'].value_counts()

df['Salary'].agg(['min', 'max', 'mean', 'median'])
```

##### Email


```python
# Fixing the email column
# Also noted the Name column header should be called 'First Name' to match 'Last Name' column header.

# Finding the number of missing values in the Email column
print(df['Email'].isna().sum())

# Renamed Name column header to First Name to match Last Name column header.
df = df.rename(columns={'Name': 'First Name'})

df['Email'] = df['First Name'].str.lower() + '.' + df['Last Name'].str.lower() + '@saltcorp.com'

df['Email'].head(10)
```

##### Benefits Enrolled

```python
#fixing the benefits enrolled column
benefits_enrolled_map = {
    'No':'No',
    'Yes':'Yes',
    'no':'No',
    'YES':'Yes',
    'yes':'Yes',
    'Y':'Yes',
    'N':'No'
}

df['Benefits Enrolled'] = df['Benefits Enrolled'].str.strip().map(benefits_enrolled_map)

df['Benefits Enrolled'] = df['Benefits Enrolled'].fillna('')

df['Benefits Enrolled'].value_counts()
```

##### Joining Date


```python
# Fixing the Join Date column
# Converting the Join Date column to a datetime data type

# Number of missing dates in the Join Date column 
# print(df['Joining Date'].isna().sum()) # -> 0

# Data type of the Join Date column  
print(df['Joining Date'].dtype) # -> string

df['Joining Date'] = pd.to_datetime(df['Joining Date'], format='mixed', errors='coerce')

df['Joining Date'] = df['Joining Date'].dt.strftime('%d/%m/%Y')

print(df['Joining Date'].dtype) # -> object

print(df['Joining Date'].isna().sum()) # -> 0
```

##### Last Promotion Date

```python
# Fixing Last Promotion Date column
# Converting the Last Promotion Date column to a datetime data type

print(df['Last Promotion Date'].dtype)
print(df['Last Promotion Date'].isna().sum()) # -> 0

# It replaced the N/A with blanks 
df['Last Promotion Date'] = pd.to_datetime(df['Last Promotion Date'], format='mixed', errors='coerce')
df['Last Promotion Date'] = df['Last Promotion Date'].dt.strftime('%d/%m/%Y')

print(df['Last Promotion Date'].dtype) # -> object
print(df['Last Promotion Date'].isna().sum()) # -> 0

df['Last Promotion Date'].value_counts().sum() # The number we are working with for the column 
```
